Q3. Write a program that takes marks of three subjects and prints the average and grade.


```python
m1 = float(input("Enter marks of subject 1: "))
m2 = float(input("Enter marks of subject 2: "))
m3 = float(input("Enter marks of subject 3: "))

avg = (m1 + m2 + m3) / 3

print("Average =", avg)

if avg >= 90:
    print("Grade A")
elif avg >= 75:
    print("Grade B")
elif avg >= 60:
    print("Grade C")
elif avg >= 50:
    print("Grade D")
else:
    print("Grade F")
```

Q13. Write a program to reverse a given number.


```python
n = int(input("Enter a number: "))

rev = 0

while n > 0:
    digit = n % 10
    rev = rev * 10 + digit
    n = n // 10

print("Reverse =", rev)
```

Q14. Write a program to check whether a number is a palindrome.

```python
n = int(input("Enter a number: "))

original = n
rev = 0

while n > 0:
    digit = n % 10
    rev = rev * 10 + digit
    n = n // 10

if original == rev:
    print("Palindrome")
else:
    print("Not Palindrome")
```

Q15. Write a program to print the Fibonacci series up to N terms.


```python
n = int(input("Enter number of terms: "))

a = 0
b = 1

for i in range(n):
    print(a, end=" ")
    a, b = b, a + b
```

Q18. Write a program to check whether a number is an Armstrong number.


```python
n = int(input("Enter a number: "))

original = n
sum = 0

while n > 0:
    digit = n % 10
    sum = sum + digit ** 3
    n = n // 10

if sum == original:
    print("Armstrong number")
else:
    print("Not an Armstrong number")
```

Q19. Write a program to print all prime numbers between 1 and N.


```python
n = int(input("Enter N: "))

for num in range(2, n + 1):
    prime = True

    for i in range(2, num):
        if num % i == 0:
            prime = False
            break

    if prime:
        print(num, end=" ")
```

Q20. Write a program to find the greatest common divisor (GCD) of two numbers

```python
a = int(input("Enter first number: "))
b = int(input("Enter second number: "))

while b != 0:
    a, b = b, a % b

print("GCD =", a)
```

Q32. Write a program to count even and odd numbers in a list.


```python
numbers = [1, 2, 3, 4, 5, 6]

even = 0
odd = 0

for n in numbers:
    if n % 2 == 0:
        even += 1
    else:
        odd += 1

print("Even =", even)
print("Odd =", odd)
```

Q34. Write a program to sort a list without using the sort() method.


```python
numbers = [5, 2, 8, 1, 3]

for i in range(len(numbers)):
    for j in range(i + 1, len(numbers)):
        if numbers[i] > numbers[j]:
            numbers[i], numbers[j] = numbers[j], numbers[i]

print("Sorted list:", numbers)
```
