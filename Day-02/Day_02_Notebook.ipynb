# ***import matplotlib.pyplot as plt***

```python
import matplotlib.pyplot as plt

x = [1,2,3,4,5]
y = [2,4,6,8,10]

plt.plot(x,y)   #Create an line graph
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.title("Line Graph")

plt.show()      #Show or Display the line graph
```

```python
from matplotlib.lines import lineMarkers
x1 = [1,2,3,4,5,6,7]
y1 = [2,4,5,8,5,9,10]
y2 = [1,2,3,4,5,6,7]


plt.plot(x1,y1, marker='o', linestyle='dashed',label = "Line1") #line Color green, linestyle for formating the line, marker for dots
plt.plot(x1,y2, color="green", marker='o', linestyle='dashed',label= "Line2")
plt.title("Green Line Graph")
plt.show()
```

# ***import numpy as np***

```python
import numpy as np

import matplotlib.pyplot as plt
import numpy as np

subjects = ["Python", "DBMS", "CN", "OS"]

boys = [80, 75, 85, 78]
girls = [85, 82, 90, 88]

x = np.arange(len(subjects))
width = 0.4

plt.bar(x - width/2, boys, width, label="Boys")
plt.bar(x + width/2, girls, width, label="Girls")

plt.xticks(x, subjects)
plt.xlabel("Subjects")
plt.ylabel("Marks")
plt.title("Subject-wise Marks")

plt.legend()
plt.show()
```

```python
#HistroGraph

import matplotlib.pyplot as plt

marks = [45, 50, 55, 60, 60, 65, 70, 70, 70, 75, 80, 85, 90]

plt.hist(marks, bins=5, edgecolor="black")

plt.title("Marks Distribution")
plt.xlabel("Marks")
plt.ylabel("Frequency")

plt.show()
```

```python
x01 = [5,6,7,8]
y01 = [80,82,85,90]

x02 = [6,7,8,9]
y02 = [60,62,65,70]

plt.scatter(x01,y01, color="blue", label="Group 1")
plt.scatter(x02,y02, color="orange", label="Group 2")

plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.title("Scatter Plot")

plt.legend()
plt.show()
```

```python
import matplotlib.pyplot as plt

subjects = ["Python", "DBMS", "CN", "OS", "ADA"]
marks = [85, 78, 90, 82, 80]

plt.pie(marks, labels=subjects, autopct="%1.1f%%", explode=[0.1,0,0,0.1,0])

plt.title("Subject-wise Marks")

plt.show()
```
